```python

sparse_0_path = Path('/home/david/projects/embodied_gaussians/datasets/real/multiple1_aruco/modelling/static/sparse_5/0')

# Load the point cloud
pts3d = np.load(sparse_0_path / 'points3D_all.npy')
col = np.load(sparse_0_path / 'pointsColor_all.npy')


```

```python
pts3d.reshape(-1, 3)
```

```python
# Create an Open3D point cloud
pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(pts3d.reshape(-1, 3))
pcd.colors = o3d.utility.Vector3dVector(col.reshape(-1, 3))

# Visualize the point cloud
o3d.visualization.draw_geometries([pcd])
```
